```python
# Carga del dataset para iniciar la Guía 2 y creación de copias de respaldo
df = pd.read_csv(ruta_final_csv)
df_original = df.copy()
df_limpio = df.copy()

print(f"Dataset Cargado: {ruta_final_csv}")
print(f"Dimensiones: {df.shape[0]} filas, {df.shape[1]} columnas")
print("Copias de trabajo creadas: -------------------")
print(f"Original: {df_original.shape}")
print(f"Para Limpiar: {df_limpio.shape}")
```

```python
# INSPECCION DEL DATASET

# Verificando tipos de datos y detalles de las columnas cargadas
print("Nombre de columnas:--------------")
print(df_limpio.columns.tolist())

print("\nTipos de datos:--------------")
print(df_limpio.dtypes)

print("\nInformacion general: ----------------")
df_limpio.info()
print("Inspección de datos completada.")
```

```python
# CONTEO DE NULOS Y DUPLICADOS

# Comprobación de la integridad inicial buscando datos faltantes y repetidos
print("\nValores Nulos: ----------")
print(df_limpio.isnull().sum())

print("\nPorcentajes de nulos: --------------")
print((df_limpio.isnull().mean() * 100).round(2))

print("\nRegistros duplicados: -----------------")
duplicados = df_limpio.duplicated().sum()
print(f"Registros duplicados: {duplicados}")
print("Verificación de nulos y duplicados finalizada.")
```

```python
# EXTRACCION DE VALORES UNICOS

# Revisión de datos categóricos para encontrar inconsistencias o errores de escritura
print("Valores unicos de categorias: -------")
columnas_categoricas = df_limpio.select_dtypes(include=['object']).columns

for columna in columnas_categoricas:
    print(f"\n{columna}:")
    print(df_limpio[columna].unique())

print("\nExtracción de valores únicos finalizada.")
```

```python
# ERRORES CONTROLADOS: INSERCION (PROPORCIONAL A 10,000 FILAS)

# Modificamos intencionalmente bloques de registros para probar la limpieza
df_errores = df_limpio.copy()

# Error 1: Espacios y Mayúsculas/Minúsculas en facultad (Filas 0 a 149)
df_errores.loc[0:49, 'facultad'] = ' ingenieria '
df_errores.loc[50:99, 'facultad'] = 'DISEÑO'
df_errores.loc[100:149, 'facultad'] = 'ciencias '

# Error 2: Valores nulos en estado de mantenimiento (Filas 200 a 299 -> 100 nulos)
df_errores.loc[200:299, 'estado'] = np.nan

# Error 3: Estudiantes negativos (Filas 400 a 449 -> 50 valores con signo negativo)
df_errores.loc[400:449, 'estudiantes_presentes'] = -15

# Error 4: Registros duplicados (Duplicamos las primeras 50 filas)
df_errores = pd.concat([df_errores, df_errores.iloc[0:50]], ignore_index=True)

print("Errores controlados insertados en el dataset de prueba (10,000+ registros).")
```

```python
# ERRORES CONTROLADOS: GUARDADO

# Exportación del dataset corrupto a la carpeta de procesados
print("-------- Dataset con errores ----------")
print(df_errores[['id_registro', 'facultad', 'estado', 'estudiantes_presentes']])

ruta_errores = DATA_PROCESSED / "dataset_con_errores_Smart_Campus.csv"
df_errores.to_csv(ruta_errores, index=False, encoding='utf-8')

print(f"\nDataset con errores guardado: {ruta_errores}")
```

```python
# LIMPIEZA DE DATOS: COLUMNAS

# Normalización de los encabezados de las columnas (minúsculas y eliminación de espacios)
df_limpio = df_errores.copy()

print("Nombres de columnas Antes: -----------------")
print(df_limpio.columns.tolist())

df_limpio.columns = (
    df_limpio.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_', regex=False)
)

print("\nNombres de columnas Despues: -----------------")
print(df_limpio.columns.tolist())
print("Limpieza de nombres de columnas ejecutada.")
```

```python
# LIMPIEZA DE DATOS: ESPACIOS EN TEXTO

# Se eliminan espacios al principio y final de los strings usando strip()
columnas_texto = df_limpio.select_dtypes(include=['object']).columns
print(f"Columnas de texto identificadas: {columnas_texto.to_list()}")

for columna in columnas_texto:
    df_limpio[columna] = df_limpio[columna].astype('string').str.strip()

print("Espacios eliminados de todas las columnas de texto.")
```

```python
# LIMPIEZA DE DATOS: CORRECCION DE FACULTAD

# Homologación de facultades utilizando minúsculas y mapeo predefinido
print("Facultad - Antes:", df_limpio['facultad'].unique().tolist())

df_limpio['facultad'] = df_limpio['facultad'].str.lower()
mapeo_facultades = {
    'ingenieria': 'Ingenieria',
    'diseño': 'Diseño',
    'ciencias': 'Ciencias'
}
df_limpio['facultad'] = df_limpio['facultad'].replace(mapeo_facultades)

print("\nFacultad - Después:", df_limpio['facultad'].unique().tolist())
print("Valores de facultad mapeados y normalizados.")
```

```python
# LIMPIEZA DE DATOS: ESTADOS Y PRIORIDADES

# Formato de título para ciertas columnas de texto y manejo seguro de valores nulos
columnas_normalizar = ['tipo_recurso', 'estado', 'prioridad']
for columna in columnas_normalizar:
    if columna in df_limpio.columns:
        df_limpio[columna] = df_limpio[columna].astype('string').str.strip().str.title()
        print(f"{columna} normalizada: {df_limpio[columna].unique().tolist()}")

df_limpio = df_limpio.replace({'nan': np.nan, '<NA>': np.nan, 'None': np.nan})

print("\nValores nulos despues de restaurar: -----------------")
print(df_limpio[['estado']].isnull().sum())

# Se rellenan nuevamente tras la manipulación
df_limpio['estado'] = df_limpio['estado'].fillna('Operativo')
df_limpio['prioridad'] = df_limpio['prioridad'].fillna('Ninguna')

print("\nEstados de mantenimiento completados -----------")
print(df_limpio[['estado', 'prioridad']].head(7))
print("Limpieza de estados y prioridades terminada.")
```

```python
# LIMPIEZA DE DATOS: DUPLICADOS Y FECHAS

# Remoción de filas duplicadas y formato datetime para las fechas
print(f"Duplicados antes: {df_limpio.duplicated().sum()}")

df_limpio = df_limpio.drop_duplicates().reset_index(drop=True)

print(f"Duplicados despues de eliminar: {df_limpio.duplicated().sum()}")
print(f"Nuevas dimensiones: {df_limpio.shape}")

df_limpio['fecha'] = pd.to_datetime(df_limpio['fecha'], errors='coerce')

print(f"Fecha convertida: {df_limpio['fecha'].dtype}")
print("Tratamiento de duplicados y conversión de fechas lista.")
```

```python
# CONVERSION DE COLUMNAS NUMERICAS

# Asegurando que las columnas cuantitativas no se traten como texto
columnas_numericas = [
    'horas_uso', 'estudiantes_presentes', 'capacidad_maxima',
    'horas_estudiante_totales', 'porcentaje_ocupacion'
]

for columna in columnas_numericas:
    if columna in df_limpio.columns:
        df_limpio[columna] = pd.to_numeric(df_limpio[columna], errors='coerce')

print("Columnas numéricas convertidas: -------")
print(df_limpio[columnas_numericas].dtypes)
print("Transformación a formatos numéricos finalizada.")
```

```python
# CORRECCION DE VALORES NEGATIVOS

# Búsqueda de valores atípicos y uso de valor absoluto para corregirlos
asistencia_negativa = df_limpio[df_limpio['estudiantes_presentes'] < 0]
print(f"Registros negativos encontrados: {len(asistencia_negativa)}")

if len(asistencia_negativa) > 0:
    print(asistencia_negativa[['id_registro', 'estudiantes_presentes']])

df_limpio['estudiantes_presentes'] = df_limpio['estudiantes_presentes'].abs()

print("Valores negativos corregidos con éxito.")
```

```python
# GUARDADO Y DESCARGA DEL DATASET LIMPIO

# Exportación del dataset limpio a la carpeta de procesados para usarlo en PySpark (Guia 3)
ruta_limpia_csv = DATA_PROCESSED / "dataset_limpio_Smart_Campus.csv"
df_limpio.to_csv(ruta_limpia_csv, index=False, encoding='utf-8')

print(f"Dataset limpio guardado exitosamente en: {ruta_limpia_csv}")

# Descarga del archivo localmente
from google.colab import files
files.download(str(ruta_limpia_csv))
print("Descarga del archivo limpio iniciada.")
```
