```python
# CREACION DE TABLA RECURSOS
conexion = sqlite3.connect(DATA_RAW / "SmartCampus_Analytics.db")
cursor = conexion.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS recursos(
    id_recurso INTEGER PRIMARY KEY,
    sede TEXT NOT NULL,
    nombre_recurso TEXT NOT NULL,
    edificio TEXT NOT NULL,
    tipo_recurso TEXT NOT NULL,
    capacidad_maxima INTEGER
)
""")
conexion.commit()

# Insertamos
df_recursos_maestro[["id_recurso", "sede", "nombre_recurso", "edificio", "tipo_recurso", "capacidad_maxima"]].to_sql(
    "recursos", conexion, if_exists="replace", index=False
)

df_recursos = pd.read_sql_query("SELECT * FROM recursos;", conexion)
print(f"Recursos en SQLite: {len(df_recursos)} edificios en las 4 sedes.")
```

```python
# CONSULTA Y CREACION DE DATAFRAME DE RECURSOS

# Aqui se realiza una consulta SQL para extraer todos los registros de la tabla de recursos alojada en la base de datos
# y cargarlos en un nuevo dataframe de Pandas para facilitar su posterior manipulacion

df_recursos = pd.read_sql_query("SELECT * FROM recursos;", conexion)

print("\nRecursos en la tabla de datos:\n")
print(df_recursos)
print("Consulta de recursos transformada en DataFrame exitosamente.")
```

```python
# INTEGRACION DE DATAFRAMES

# Aqui se procede a fusionar el dataframe de asistencia original con el nuevo dataframe de recursos
# utilizando la clave comun 'id_recurso' para integrar toda la informacion descriptiva en una sola estructura

df_integrado = pd.merge(df_asistencia, df_recursos, on="id_recurso", how="left")

print("\nAsistencia con informacion de recursos:\n")
print(df_integrado[["id_registro", "nombre_recurso", "tipo_recurso", "hora_inicio", "hora_fin", "estudiantes_presentes", "capacidad_maxima"]])
print("Dataframes integrados correctamente.")
```

```python
# CALCULO DE PORCENTAJE DE OCUPACION

# Aqui se calcula el porcentaje de ocupacion dividiendo la cantidad de estudiantes presentes entre
# la capacidad maxima de cada recurso, agregando este resultado como una nueva columna al dataframe integrado

df_integrado["porcentaje_ocupacion"] = round((df_integrado["estudiantes_presentes"] / df_integrado["capacidad_maxima"]) * 100, 2)

print("\nAsistencia con porcentaje de ocupacion de los laboratorios:")
print(df_integrado[["id_registro", "nombre_recurso", "estudiantes_presentes", "capacidad_maxima", "porcentaje_ocupacion"]])
print("Calculo de porcentaje de ocupacion completado.")
```

```python
# CREACION DE DATOS DE MANTENIMIENTO

# Aqui se crea una lista estructurada mediante diccionarios para almacenar la informacion detallada sobre los
# mantenimientos programados, incluyendo el estado, la prioridad y la facultad asignada a cada recurso

mantenimientos = [
    {"id_recurso": 201, "estado": "Mantenimiento Preventivo", "prioridad": "Media", "facultad": "Ingenieria"},
    {"id_recurso": 202, "estado": "Actualizacion Software", "prioridad": "Alta", "facultad": "Diseño"},
    {"id_recurso": 204, "estado": "Reemplazo de Hardware", "prioridad": "Alta", "facultad": "Ingenieria"}
]

print("Datos de mantenimientos programados:")
for m in mantenimientos:
    print(f"  Recurso {m['id_recurso']}: {m['estado']} (Prioridad {m['prioridad']}) para {m['facultad']}")
print("Estructura de mantenimientos generada.")
```

```python
# GUARDADO DE DATOS EN FORMATO JSON

# Aqui se exporta la informacion detallada de los mantenimientos programados hacia un archivo estructurado
# en formato json, almacenandolo especificamente en el directorio de datos externos del proyecto

ruta_json = DATA_EXTERNAL / "mantenimientos.json"
with open(ruta_json, "w", encoding="utf-8") as archivo:
    json.dump(mantenimientos, archivo, ensure_ascii=False, indent=4)

print(f"Archivo json guardado en: {ruta_json}")
```

```python
# LECTURA DE ARCHIVO JSON

# Aqui se procede a leer el archivo json que fue previamente guardado utilizando la libreria pandas,
# con el objetivo de cargar esta informacion en un nuevo dataframe independiente para su analisis

df_mantenimientos = pd.read_json(ruta_json)

print("Mantenimientos con DataFrame:")
print(df_mantenimientos)
print("Archivo JSON cargado exitosamente.")
```

```python
# INTEGRACION FINAL Y MANEJO DE NULOS

# Aqui se realiza la union definitiva de los datos integrados con la informacion de los mantenimientos correspondientes,
# ademas de llevar a cabo una limpieza al rellenar los valores nulos en el estado y la prioridad para garantizar
# que el dataset resultante este completamente consolidado
df_final = pd.merge(df_integrado, df_mantenimientos, on=["id_recurso", "facultad"], how="left")
df_final["estado"] = df_final["estado"].fillna("Operativo")
df_final["prioridad"] = df_final["prioridad"].fillna("Ninguna")

print("\nDataset final integrado con estados de mantenimiento:\n")
print(df_final.head())
print("Nulos rellenados y consolidación finalizada.")
```

```python
# ERRORES CONTROLADOS: INSERCION (PROPORCIONAL A 10,000 FILAS)

# Modificamos intencionalmente bloques de registros para probar la limpieza
df_errores = df_limpio.copy()

# Error 1: Espacios y Mayúsculas/Minúsculas en facultad (Filas 0 a 149)
df_errores.loc[0:49, 'facultad'] = ' ingenieria '
df_errores.loc[50:99, 'facultad'] = 'DISEÑO'
df_errores.loc[100:149, 'facultad'] = 'ciencias '

# Error 2: Valores nulos en estado de mantenimiento (Filas 200 a 299 -> 100 nulos)
df_errores.loc[200:299, 'estado'] = np.nan

# Error 3: Estudiantes negativos (Filas 400 a 449 -> 50 valores con signo negativo)
df_errores.loc[400:449, 'estudiantes_presentes'] = -15

# Error 4: Registros duplicados (Duplicamos las primeras 50 filas)
df_errores = pd.concat([df_errores, df_errores.iloc[0:50]], ignore_index=True)

print("Errores controlados insertados en el dataset de prueba (10,000+ registros).")
```

```python
# ERRORES CONTROLADOS: GUARDADO

# Exportación del dataset corrupto a la carpeta de procesados
print("-------- Dataset con errores ----------")
print(df_errores[['id_registro', 'facultad', 'estado', 'estudiantes_presentes']])

ruta_errores = DATA_PROCESSED / "dataset_con_errores_Smart_Campus.csv"
df_errores.to_csv(ruta_errores, index=False, encoding='utf-8')

print(f"\nDataset con errores guardado: {ruta_errores}")
```

```python
# LIMPIEZA DE DATOS: COLUMNAS

# Normalización de los encabezados de las columnas (minúsculas y eliminación de espacios)
df_limpio = df_errores.copy()

print("Nombres de columnas Antes: -----------------")
print(df_limpio.columns.tolist())

df_limpio.columns = (
    df_limpio.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_', regex=False)
)

print("\nNombres de columnas Despues: -----------------")
print(df_limpio.columns.tolist())
print("Limpieza de nombres de columnas ejecutada.")
```

```python
# LIMPIEZA DE DATOS: ESPACIOS EN TEXTO

# Se eliminan espacios al principio y final de los strings usando strip()
columnas_texto = df_limpio.select_dtypes(include=['object']).columns
print(f"Columnas de texto identificadas: {columnas_texto.to_list()}")

for columna in columnas_texto:
    df_limpio[columna] = df_limpio[columna].astype('string').str.strip()

print("Espacios eliminados de todas las columnas de texto.")
```

```python
# LIMPIEZA DE DATOS: CORRECCION DE FACULTAD

# Homologación de facultades utilizando minúsculas y mapeo predefinido
print("Facultad - Antes:", df_limpio['facultad'].unique().tolist())

df_limpio['facultad'] = df_limpio['facultad'].str.lower()
mapeo_facultades = {
    'ingenieria': 'Ingenieria',
    'diseño': 'Diseño',
    'ciencias': 'Ciencias'
}
df_limpio['facultad'] = df_limpio['facultad'].replace(mapeo_facultades)

print("\nFacultad - Después:", df_limpio['facultad'].unique().tolist())
print("Valores de facultad mapeados y normalizados.")
```

```python
# LIMPIEZA DE DATOS: ESTADOS Y PRIORIDADES

# Formato de título para ciertas columnas de texto y manejo seguro de valores nulos
columnas_normalizar = ['tipo_recurso', 'estado', 'prioridad']
for columna in columnas_normalizar:
    if columna in df_limpio.columns:
        df_limpio[columna] = df_limpio[columna].astype('string').str.strip().str.title()
        print(f"{columna} normalizada: {df_limpio[columna].unique().tolist()}")

df_limpio = df_limpio.replace({'nan': np.nan, '<NA>': np.nan, 'None': np.nan})

print("\nValores nulos despues de restaurar: -----------------")
print(df_limpio[['estado']].isnull().sum())

# Se rellenan nuevamente tras la manipulación
df_limpio['estado'] = df_limpio['estado'].fillna('Operativo')
df_limpio['prioridad'] = df_limpio['prioridad'].fillna('Ninguna')

print("\nEstados de mantenimiento completados -----------")
print(df_limpio[['estado', 'prioridad']].head(7))
print("Limpieza de estados y prioridades terminada.")
```

```python
# LIMPIEZA DE DATOS: DUPLICADOS Y FECHAS

# Remoción de filas duplicadas y formato datetime para las fechas
print(f"Duplicados antes: {df_limpio.duplicated().sum()}")

df_limpio = df_limpio.drop_duplicates().reset_index(drop=True)

print(f"Duplicados despues de eliminar: {df_limpio.duplicated().sum()}")
print(f"Nuevas dimensiones: {df_limpio.shape}")

df_limpio['fecha'] = pd.to_datetime(df_limpio['fecha'], errors='coerce')

print(f"Fecha convertida: {df_limpio['fecha'].dtype}")
print("Tratamiento de duplicados y conversión de fechas lista.")
```

```python
# CONVERSION DE COLUMNAS NUMERICAS

# Asegurando que las columnas cuantitativas no se traten como texto
columnas_numericas = [
    'horas_uso', 'estudiantes_presentes', 'capacidad_maxima',
    'horas_estudiante_totales', 'porcentaje_ocupacion'
]

for columna in columnas_numericas:
    if columna in df_limpio.columns:
        df_limpio[columna] = pd.to_numeric(df_limpio[columna], errors='coerce')

print("Columnas numéricas convertidas: -------")
print(df_limpio[columnas_numericas].dtypes)
print("Transformación a formatos numéricos finalizada.")
```

```python
# IMPORTACION DE LIBRERIAS

# Aqui se importan las librerias necesarias como SparkSession y functions de pyspark,
# ademas de pandas y pathlib para manejar las rutas y los cruces de informacion
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pathlib import Path
import pandas as pd

print("Librerias importadas ---------------")
```

```python
# CREACION DE SESION DE SPARK

# Aqui se inicializa y configura la sesion local de Spark, estableciendo
# el nombre de la aplicacion y definiendo que se usaran todos los nucleos locales disponibles
spark = SparkSession.builder \
  .master("local[*]")\
  .appName("SmartCampusAI_Guia3")\
  .getOrCreate()

print(f"Spark Version: {spark.version}")
print(f"Paralelismo: {spark.sparkContext.defaultParallelism}")
print("Sesion de spark inicializada correctamente :C")
```

```python
# DEFINICION DE RUTAS DE TRABAJO

# Aqui se configuran las rutas de los directorios tanto para leer los datos procesados
# en la guia 2 como para guardar los resultados analiticos que genere Spark
BASE_DIR = Path("/content")
DATA_PROCESSED = BASE_DIR/"datos"/"procesados"
DATA_ANALYTICS = BASE_DIR/"datos"/"analiticos"

DATA_PROCESSED.mkdir(parents=True, exist_ok=True)
DATA_ANALYTICS.mkdir(parents=True, exist_ok=True)

print(f"Datos procesados: {DATA_PROCESSED}")
print(f"Datos analiticos: {DATA_ANALYTICS}")
print("Rutas definidas y carpetas creadas :C")
```

```python
# CARGA DE DATASET LIMPIO

# Aqui se procede a leer el archivo csv limpio generado en la guia anterior y
# se carga en un dataframe de Spark infiriendo el esquema automaticamente
ruta_entrada = DATA_PROCESSED/"dataset_limpio_Smart_Campus.csv"

print(f"Leyendo: {ruta_entrada}")

df = spark.read\
  .option("header", True)\
  .option("inferSchema", True)\
  .csv(str(ruta_entrada))

print(f"Filas: {df.count()}")
print(f"Columnas: {len(df.columns)}")
print("Dataset cargado en Spark exitosamente :C")
```

```python
# VERIFICACION DE ESQUEMA Y DATOS

# Aqui se imprime la estructura del dataframe inferida por Spark y
# se muestran las primeras filas para comprobar que los datos esten correctos
print("\nEsquema de datos: -------------:c")
df.printSchema()

print("\nPrimeras filas: ")
df.show(5, truncate= False)
print("Esquema y datos verificados :C")
```

```python
# INSPECCION DE PARTICIONES

# Aqui se verifica el numero de particiones iniciales en las que Spark
# ha dividido el dataframe al momento de realizar la lectura
print(f"\nParticiones iniciales: {df.rdd.getNumPartitions()}")
print("Inspeccion de particiones completada :C")
```

```python
# FILTRADO DE REGISTROS VALIDOS

# Aqui se filtran los registros del dataframe conservando unicamente
# aquellos donde los estudiantes presentes en el laboratorio son mayores a cero
df_validos = df.filter(F.col("estudiantes_presentes") > 0)

print(f"Registros despues de filtrar: {df_validos.count()}")
print(f"Registros eliminados: {df.count() - df_validos.count()}")
print("Filtrado de registros finalizado :C")
```

```python
# CREACION DE COLUMNA DERIVADA

# Aqui se crea una nueva columna llamada espacios_disponibles restando
# la cantidad de estudiantes presentes de la capacidad maxima del recurso
df_enriquecido = df_validos.withColumn(
    "espacios_disponibles",
    F.col("capacidad_maxima") - F.col("estudiantes_presentes")
)

print("Columna 'espacios_disponibles' creada: --------:c")
df_enriquecido.select("id_registro", "capacidad_maxima", "estudiantes_presentes", "espacios_disponibles").show(5)
print("Columna derivada calculada y agregada correctamente :C")
```

```python
# AGRUPACION POR FACULTAD

# Aqui se agrupan los datos por facultad para calcular el uso total de horas,
# la ocupacion promedio y el total de sesiones registradas utilizando funciones de Spark
resumen_facultad = df_enriquecido.groupBy("facultad").agg(
  F.sum("horas_estudiante_totales").alias("uso_total_horas"),
  F.avg("porcentaje_ocupacion").alias("ocupacion_promedio"),
  F.count("*").alias("sesiones_registradas")
).orderBy(F.desc("uso_total_horas"))

print("Uso por facultad: -----------------:c")
resumen_facultad.show(truncate=False)
print("Agrupacion por facultad completada :C")
```

```python
# AGRUPACION POR TIPO DE RECURSO

# Aqui se realiza una agregacion similar pero agrupando por tipo de recurso
# para obtener metricas de uso especificas de laboratorios, infraestructuras, etc.
resumen_recurso = df_enriquecido.groupBy("tipo_recurso").agg(
  F.sum("horas_estudiante_totales").alias("uso_total_horas"),
  F.count("*").alias("sesiones_registradas")
).orderBy(F.desc("uso_total_horas"))

print("Uso por tipo de recurso: -----------------:c")
resumen_recurso.show(truncate=False)
print("Agrupacion por recurso completada :C")
```

```python
# CREACION DE DIMENSION Y CRUCE DE DATOS

# Aqui se define un nuevo dataframe categorico sobre la criticidad de los recursos
# y se une (join) con nuestro dataframe principal usando el tipo_recurso como llave
tipos_rec = [
    ("Laboratorio", "Fisico", "Alta"),
    ("Equipamiento", "Hardware", "Media"),
    ("Infraestructura", "Servidor", "Critica")
]

df_tipos = spark.createDataFrame(tipos_rec, ["tipo_recurso", "categoria_recurso", "criticidad"])

print("Tabla de dimension de recursos: -----------------:c")
df_tipos.show()

df_con_recurso = df_enriquecido.join(df_tipos, on="tipo_recurso", how="left")

print(f"Filas antes del join: {df_enriquecido.count()}")
print(f"Filas despues del join: {df_con_recurso.count()}")

print("Datos combinados con detalles del recurso: -----------------:c")
df_con_recurso.select("id_registro", "tipo_recurso", "categoria_recurso", "criticidad", "horas_estudiante_totales").show(5)
print("Cruce de datos con tabla de dimension completado :C")
```

```python
# CREACION DE VISTA TEMPORAL

# Aqui se registra el dataframe procesado como una vista temporal en memoria
# para permitir realizar consultas directamente utilizando lenguaje SQL
df_con_recurso.createOrReplaceTempView("campus_analiticas")
print("Vista temporal 'campus_analiticas' creada exitosamente :C")
```

```python
# EJECUCION DE CONSULTA SQL

# Aqui se ejecuta una consulta SQL nativa sobre la vista temporal para
# extraer un resumen analitico del uso agrupado por cada tipo de recurso
consulta = """
  SELECT
    tipo_recurso,
    COUNT(*) AS sesiones_registradas,
    ROUND(SUM(horas_estudiante_totales), 2) AS uso_total_horas,
    ROUND(AVG(porcentaje_ocupacion), 2) AS ocupacion_promedio
  FROM campus_analiticas
  GROUP BY tipo_recurso
  ORDER BY uso_total_horas DESC
"""

resultado_sql = spark.sql(consulta)
print("Consulta SQL - Resumen por recurso: -----------------:c")
resultado_sql.show(truncate=False)
print("Consulta SQL ejecutada correctamente :C")
```

```python
# INSPECCION DEL PLAN DE EJECUCION

# Aqui se muestra el plan fisico y logico que Spark ha diseñado por debajo
# para resolver y optimizar la consulta SQL realizada previamente
print("\nPlan de ejecucion formateado: -----------:c")
resultado_sql.explain("formatted")
print("Plan de ejecucion desplegado con exito :C")
```

```python
# REPARTICION DE DATOS

# Aqui se evaluan las particiones actuales del dataframe y se obliga a Spark
# a reparticionarlo en 4 bloques para optimizar el rendimiento de la escritura
print(f"Particiones actuales {df_con_recurso.rdd.getNumPartitions()}")

df_particionado = df_con_recurso.repartition(4)
print(f"Nuevas particiones: {df_particionado.rdd.getNumPartitions()}")
print("Reparticion de datos completada :C")
```

```python
# GUARDADO EN FORMATO PARQUET

# Aqui se guarda el dataframe procesado en formato parquet dentro de la carpeta
# analitica, el cual es un formato de almacenamiento altamente optimizado
ruta_parquet = DATA_ANALYTICS/"campus_analiticas_parquet"

df_particionado.write \
  .mode("overwrite") \
  .parquet(str(ruta_parquet))

print(f"Parquet guardado en: {ruta_parquet}")
print("Guardado en formato Parquet finalizado :C")
```

```python
# LECTURA DE ARCHIVO PARQUET

# Aqui se lee nuevamente el archivo parquet recien creado desde el almacenamiento
# para verificar que la informacion se ha persistido correctamente
df_leido = spark.read.parquet(str(ruta_parquet))

print(f"Filas leidas desde Parquet: {df_leido.count()}")
print("Primeras filas: ------------:c")
df_leido.show(5, truncate=False)
print("Lectura de Parquet completada :C")
```

```python
# AGREGACION UTILIZANDO PANDAS

# Aqui se realiza la misma agrupacion inicial por facultad pero utilizando la
# libreria Pandas pura para posteriormente comparar los resultados con Spark
df_pd = pd.read_csv(DATA_PROCESSED/"dataset_limpio_Smart_Campus.csv")

pd_facultad = df_pd.groupby("facultad")["horas_estudiante_totales"].sum().reset_index()
pd_facultad.columns = ["facultad", "uso_total_horas"]

print("Pandas uso por facultad: -----------------:c")
print(pd_facultad)
print("Agregacion con Pandas ejecutada :C")
```

```python
# CONVERSION DE RESULTADOS DE SPARK A PANDAS

# Aqui se toma el dataframe resumido de Spark creado anteriormente y se convierte
# explicitamente a un dataframe de Pandas para facilitar la integracion
sp_facultad = resumen_facultad.select("facultad", "uso_total_horas").orderBy("facultad").toPandas()
print("\nSpark - uso por facultad: ------------------:c")
print(sp_facultad)
print("Conversion de Spark a Pandas finalizada :C")
```

```python
# COMPARACION DE RESULTADOS

# Aqui se cruzan los dataframes resultantes de Pandas y Spark haciendo un merge
# para validar que los calculos agrupados coinciden en ambas herramientas
print("\nComparacion Pandas vs Spark")

comparacion = pd.merge(pd_facultad, sp_facultad, on="facultad", suffixes=('_pandas', '_spark'))
print(comparacion)
print("Comparacion completada: Los resultados son consistentes entre pandas y spark :C")
```

```python
# CREACIÓN DE DATOS FICTICIOS PARA SMARTCAMPUS (10,000 ALUMNOS)

# Si faker no está instalado en el entorno, descomentar la siguiente línea:
# !pip install faker

from faker import Faker
import random

fake = Faker('es_ES')

N_ALUMNOS = 10000

catalogo_carreras = {
    "Ingenieria": ["Sistemas", "Industrial", "Electrica", "Mecatronica"],
    "Diseño": ["Animacion Digital", "Diseño Grafico", "Diseño de Interiores"],
    "Ciencias": ["Ciencias de la Computacion", "Matematica Aplicada"]
}

facultades_disponibles = list(catalogo_carreras.keys())

registros_estudiantes = []
for i in range(1, N_ALUMNOS + 1):
    fac = random.choice(facultades_disponibles)
    carr = random.choice(catalogo_carreras[fac])
    nombre_alumno = fake.name()

    # Sanitización simple para el correo institucional
    alias_correo = (
        nombre_alumno.lower()
        .replace(" ", ".")
        .replace("ñ", "n")
        .replace("á", "a")
        .replace("é", "e")
        .replace("í", "i")
        .replace("ó", "o")
        .replace("ú", "u")
    )[:18]

    registros_estudiantes.append({
        "id_estudiante": 20260000 + i,
        "nombre": nombre_alumno,
        "correo": f"{alias_correo}@campus.edu.sv",
        "telefono": f"7{random.randint(1000000, 9999999)}",
        "edad": int(np.random.choice([18, 19, 20, 21, 22, 23, 25, 29, 35, 42],
                                     p=[0.20, 0.25, 0.20, 0.15, 0.08, 0.05, 0.03, 0.02, 0.015, 0.005])),
        "facultad": fac,
        "carrera": carr
    })

estudiantes = pd.DataFrame(registros_estudiantes)

print(f"Datos ficticios de estudiantes generados: {len(estudiantes)} registros.")
print(estudiantes.head(3))
```

```python
# MATRIZ DE CLASIFICACIÓN DE DATOS

# Definimos el nivel de sensibilidad de cada columna basándonos en la normativa
# de protección de datos para establecer qué controles técnicos aplicar.
clasificacion = pd.DataFrame([
    ["id_estudiante", "Identificador interno", "Confidencial", "Vincular registros", "Seudonimizar"],
    ["nombre", "Identificador directo", "Confidencial", "Contacto", "Excluir de vista analítica"],
    ["correo", "Contacto", "Confidencial", "Contacto", "Enmascarar o excluir"],
    ["telefono", "Contacto", "Confidencial", "Contacto", "Enmascarar o excluir"],
    ["edad", "Demográfico", "Interno", "Segmentación", "Generalizar por rango"],
    ["facultad", "Ubicación académica", "Interno", "Análisis de uso", "Conservar"],
    ["carrera", "Académico", "Interno", "Analítica estudiantil", "Conservar"]
],
columns=["campo", "tipo", "clasificacion", "finalidad", "control"])

print("Matriz de clasificacion estructurada---------------")
print(clasificacion)
```

```python
# GUARDADO DE LA MATRIZ DE CLASIFICACIÓN

# Exportamos la matriz a un archivo Excel para que quede como evidencia
# documental de las políticas de gobierno de datos aplicadas.
ruta_clasificacion = DOCS_DIR/"clasificacion_datos_Smart_Campus.xlsx"
clasificacion.to_excel(ruta_clasificacion, index=False)
print(f"Clasificacion guardada en: {ruta_clasificacion}")
```

```python
# SEUDONIMIZACIÓN DE IDENTIFICADORES

# Aplicamos una función hash SHA-256 junto con una "sal" (salt) criptográfica
# para convertir el carnet del estudiante en un token irreversible de 12 caracteres.
SALT = "smartcampus-datos-sinteticos"
def tokenizar(valor):
    """Convierte un valor en un codigo unico de 12 caracteres."""
    texto = f"{SALT}|{valor}".encode('utf-8')
    return hashlib.sha256(texto).hexdigest()[:12]

estudiantes["estudiante_token"] = estudiantes["id_estudiante"].astype(str).apply(tokenizar)

print("Seudonimizacion aplicada correctamente:-----------")
print(estudiantes[["id_estudiante", "estudiante_token"]])
```

```python
# ENMASCARAMIENTO DE CORREO ELECTRÓNICO

# Ocultamos la información sensible del correo dejando visible solo la primera
# letra del usuario y el dominio institucional para mantener el formato.
def enmascarar_correo(correo):
    """Muestra solo la primera letra del correo y el dominio completo"""
    usuario, dominio = correo.split('@', 1)
    visible = usuario[:1]
    return f"{visible}***@{dominio}"

estudiantes["correo_mask"] = estudiantes["correo"].apply(enmascarar_correo)
print("Correo enmascarado exitosamente:----------------")
print(estudiantes[["correo", "correo_mask"]])
```

```python
# ENMASCARAMIENTO DE NÚMERO TELEFÓNICO

# Ocultamos los primeros dígitos del teléfono y mostramos solo los últimos 4
# usando una función lambda rápida y efectiva.
estudiantes["telefono_mask"] = estudiantes["telefono"].astype(str).apply(lambda x: "****" + x[-4:])
print("Telefono enmascarado exitosamente:----------------")
print(estudiantes[["telefono", "telefono_mask"]])
```

```python
# GENERALIZACIÓN DE DATOS DEMOGRÁFICOS (EDAD)

# Convertimos la edad exacta en rangos (bins) para evitar la reidentificación
# de estudiantes que puedan tener edades atípicas en ciertas carreras.
bins = [0, 24, 34, 44, 54, 64, 200]
labels = ["<=24", "25-34", "35-44", "45-54", "55-64", "65+"]

estudiantes["rango_edad"] = pd.cut(estudiantes["edad"], bins=bins, labels=labels)
print("Edad generalizada por rangos:---------------")
print(estudiantes[["edad", "rango_edad"]])
```

```python
# CREACIÓN DE VISTA ANALÍTICA PROTEGIDA (MINIMIZACIÓN)

# Aplicamos el principio de minimización de datos: extraemos únicamente las columnas
# que ya están protegidas y que son estrictamente necesarias para el análisis.
vista_analitica = estudiantes[[
    "estudiante_token",
    "rango_edad",
    "facultad",
    "carrera"
]].copy()

print("Vista analitica protegida (Sin identificadores directos):-----------")
print(vista_analitica)
```

```python
# GUÍA 6 / FASE GEOESPACIAL: MAPA DE LAS 4 SEDES NACIONALES ARCGIS

import pandas as pd
import numpy as np
from pathlib import Path
from google.colab import files

print("Generando capa para ArcGIS (4 Sedes ITCA-FEPADE)...")

GEO_DIR = BASE_DIR / "datos" / "geo"
GEO_DIR.mkdir(parents=True, exist_ok=True)

# 1. LECTURA DEL DATASET LIMPIO
df_limpio_fuente = pd.read_csv(DATA_PROCESSED / "dataset_limpio_Smart_Campus.csv")

# Asegurar tipos y recalculación directa
df_limpio_fuente["id_recurso"] = pd.to_numeric(df_limpio_fuente["id_recurso"], errors="coerce").astype(int)
df_limpio_fuente["estudiantes_presentes"] = pd.to_numeric(df_limpio_fuente["estudiantes_presentes"], errors="coerce").abs()
df_limpio_fuente["horas_uso"] = pd.to_numeric(df_limpio_fuente["horas_uso"], errors="coerce").fillna(2.0)

# Cruzar capacidad máxima desde el catálogo
df_limpio_fuente = df_limpio_fuente.drop(columns=["capacidad_maxima"], errors="ignore")
df_limpio_fuente = pd.merge(df_limpio_fuente, df_recursos_maestro[["id_recurso", "capacidad_maxima"]], on="id_recurso", how="left")

# Métricas recalculadas
df_limpio_fuente["horas_estudiante_calc"] = df_limpio_fuente["horas_uso"] * df_limpio_fuente["estudiantes_presentes"]
df_limpio_fuente["porcentaje_ocupacion_calc"] = (df_limpio_fuente["estudiantes_presentes"] / df_limpio_fuente["capacidad_maxima"]) * 100

# 2. AGREGACIÓN EQUITATIVA POR EDIFICIO
demanda_recursos = df_limpio_fuente.groupby("id_recurso").agg(
    total_sesiones=("id_registro", "count"),
    horas_estudiante_totales=("horas_estudiante_calc", "sum"),
    ocupacion_promedio=("porcentaje_ocupacion_calc", "mean"),
    total_estudiantes=("estudiantes_presentes", "sum")
).reset_index()

demanda_recursos["horas_estudiante_totales"] = demanda_recursos["horas_estudiante_totales"].round(2)
demanda_recursos["ocupacion_promedio"] = demanda_recursos["ocupacion_promedio"].round(2)

# 3. CRUCE ESPACIAL CON LAS COORDENADAS
capa_arcgis = pd.merge(df_recursos_maestro, demanda_recursos, on="id_recurso", how="left")

# 4. VALIDACIÓN TERRITORIAL PARA EL SALVADOR
mask_valida = (
    capa_arcgis['latitud'].between(13, 15) &
    capa_arcgis['longitud'].between(-91, -87)
)
print(f"Total edificios evaluados: {len(capa_arcgis)}")
print(f"Coordenadas válidas en El Salvador: {mask_valida.sum()}/20")

# 5. SEMÁFORO DE DEMANDA
capa_arcgis["alerta_demanda"] = np.where(
    capa_arcgis["ocupacion_promedio"] >= 80, "Alta Demanda (Rojo)",
    np.where(capa_arcgis["ocupacion_promedio"] >= 60, "Uso Moderado (Amarillo)", "Disponible (Verde)")
)

# 6. EXPORTACIÓN Y DESCARGA
ruta_csv_arcgis = GEO_DIR / "smartcampus_nacional_arcgis.csv"
capa_arcgis.to_csv(ruta_csv_arcgis, index=False, encoding='utf-8')

print(f"\nDescargando archivo nacional: {ruta_csv_arcgis.name}")
files.download(str(ruta_csv_arcgis))
```
